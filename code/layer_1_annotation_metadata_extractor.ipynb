```python
import os
import shutil
import pandas as pd
```

```python
CLASSES = ['Crack', 'Manhole', 'Miscellaneous abnormality', 'Pothole', 'Road patch failure', 'Surface depression']
```

```python
images = os.listdir("./Road Abnormality Dataset YOLOv11 Dataset/train/images")
labels = os.listdir("./Road Abnormality Dataset YOLOv11 Dataset/train/labels")
```

```python
os.makedirs("./Road Abnormality Dataset", exist_ok=True)
os.makedirs("./Road Abnormality Dataset/Images", exist_ok=True)
os.makedirs("./Road Abnormality Dataset/Labels", exist_ok=True)
for img, lbl in zip(images, labels):
    img_name = os.path.splitext(img)[0]
    lbl_name = os.path.splitext(lbl)[0]
    
    if img_name != lbl_name:
        print(f"Mismatch: {img} and {lbl}")
    
    else:
        name = img_name[:9]
        shutil.copy2(f"./Road Abnormality Dataset YOLOv11 Dataset/train/images/{img}", f"./Road Abnormality Dataset/Images/{name}.jpg")
        shutil.copy2(f"./Road Abnormality Dataset YOLOv11 Dataset/train/labels/{lbl}", f"./Road Abnormality Dataset/Labels/{name}.txt")
```

```python
images_dir = [f"./Road Abnormality Dataset/Images/{i}" for i in os.listdir("./Road Abnormality Dataset/Images")]
labels_dir = [f"./Road Abnormality Dataset/Labels/{i}" for i in os.listdir("./Road Abnormality Dataset/Labels")]
```

```python
df = pd.DataFrame(columns=["image_id", "abnormality_id", "abnormality_type", "x_centre", "y_centre", "width_scaled", "height_scaled"])
```

```python
img_id, abnormality_id, abnormality_type, x_centre, y_centre, width_scaled, height_scaled = [], [], [], [], [], [], []
for img, lbl in zip(images_dir, labels_dir):
    img_name = img.split("/")[-1].split(".")[0]
    lbl_name = lbl.split("/")[-1].split(".")[0]
    
    with open(lbl, "r") as f:
        lines = f.readlines()
    
    for index, line in enumerate(lines):
        class_id = int(line.split()[0])
        class_name = CLASSES[class_id]
        img_id.append(img_name)
        abnormality_id.append(f"{img_name}_annotation_{index:03d}")
        abnormality_type.append(class_name)
        x_centre.append(round(float(line.split()[1]), 4))
        y_centre.append(round(float(line.split()[2]), 4))
        width_scaled.append(round(float(line.split()[3]), 4))
        height_scaled.append(round(float(line.split()[4]), 4))

df["image_id"] = img_id
df["abnormality_id"] = abnormality_id
df["abnormality_type"] = abnormality_type
df["x_centre"] = x_centre
df["y_centre"] = y_centre
df["width_scaled"] = width_scaled
df["height_scaled"] = height_scaled
```
